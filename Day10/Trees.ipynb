```python
from collections import deque
def level_order(root):
  if root is None:
    return
  q=deque([root])
  while q:
    node=q.popleft()
    print(node.val,end=" ")
    if node.left:
      q.append(node.left)
    
    if node.right:
      q.append(node.right)
print("Level Order:")
# level_order(root)
```

```python
# count nodes
def count(root):
  if root is None:
    return 0
  return 1 + count(root.left) + count(root.right)

cnt = count(root)
print(cnt)
```

```python
# build a tree using level_order
#      1
#     /\
#    2  3
#   /    \
#  4      5
# level= 1 2 3 4 null null 5 

def buildTree(level):
  if level==[]:
    return None
  root=Tree(level[0])
  q=deque([root])
  i=1
  while q and i<len(level):
    cur=q.popleft()
    if i<len(level):
      if level[i] is not None:
        cur.left=Tree(level[i])
        q.append(cur.left)
      i+=1
    if i<len(level):
      if level[i] is not None:
        cur.right=Tree(level[i])
        q.append(cur.right)
      i+=1
  return root
level=[1,2,3,4,None,None,5]
root=buildTree(level)
level_order(root)
```

```python
# inorder
def inorder(root):
  if root is None:
    return
  inorder(root.left)
  print(root.val,end=" ")
  inorder(root.right)
```

```python

def insertBST(root,val):
  if root is None:
    return Tree(val)
  if root.val>val:
    root.left=insertBST(root.left,val)
  elif root.val<val:
    root.right=insertBST(root.right,val)
  return root

def inordersucc(root):
  while root and root.left:
    root=root.left
  return root
def deletebst(root,val):
  if root is None:
    return None
  if root.val>val:
    root.left=deletebst(root.left,val)
  elif root.val<val:
    root.right=deletebst(root.right,val)
  else:
    if not root.left:
      node=root.right
      del root
      return node
    elif not root.right:
          node=root.left
          del root
          return node
    else:
      succ=inordersucc(root.right)
      root.val=succ.val
      root.right=deletebst(root.right,succ.val)
  return root
    
def buildbst(l):
  if l==[]:
    return None
  root=None
  i=0
  while i<len(l):
    root=insertBST(root,l[i])
    i+=1
  return root

nodes=[1,2,43,4,56,6]
root=buildbst(nodes)
inorder(root)
```
