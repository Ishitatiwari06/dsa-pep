Hashing- process of storing data in key value pair for faster retrieval
Hash map- data structures to implement hashing
Hash table - table used to store key-val pairs
Collision - occurs when 2 values at same key
to prevent collision - chaining, open addressing
chaining - using ll
open addressing - linear and quadratic probing 
linear probing - checks for the next available slot sequentially when collision occurs
quadratic probing - skips slots using quadratic func when collision occurs

double hashing - open addressing collision resolution technique that uses 2 distinct hash func
load factor - no. of elems / table size
            - higher lf, higher collisions if >0.75 reshaping is suggested
Rehashing - resizing the hashtables when load factor is too high by creating new array (having double the original size often prime no.)
