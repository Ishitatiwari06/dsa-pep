```python
# i is typecode - for signed int
# I for unsigned int
# others are d,b,f
import array as arr
a=arr.array('i',[1,2,3,4,5])
```

```python
for i in a:
  print(i)
```

```python
b=a.reverse()
print(b)
```

```python
# sum of max and min elem
print(a)
mini=a[0]
maxi=a[0]
for i in a:
  if(i<mini):
    mini=i
  if(i>maxi):
    maxi=i
print(mini+maxi)
```
