```python
# buuble sort
def bubble(arr):
  n=len(arr)
  for i in range(n):
    issort=False
    for j in range(1,n-i):
      if(arr[j-1]>arr[j]):
        arr[j],arr[j-1]=arr[j-1],arr[j]
        issort=True
    if not issort:
      break
    
arr=[4,5,62,7,2,7,1]
print("Before Sort")
print(arr)
bubble(arr)
print("After Sort")
print(arr)
      
```

```python
# selection sort
def selection(arr):
  n=len(arr)
  for i in range(n):
    mini=arr[i]
    for j in range(i+1,n):
      if mini>arr[j]:
        mini=arr[j]
    arr[mini],arr[j]=arr[j],arr[mini]
    
arr=[4,5,62,7,2,7,1]
print("Before Sort")
print(arr)
bubble(arr)
print("After Sort")
print(arr)
```
