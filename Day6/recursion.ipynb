```python
def printno(i,n):
  if n<i:
    return
  print(i)
  printno(i+1,n)
printno(1,5)
```

```python
def sumofn(n):
  if n==0:
    return 0
  s = n + sumofn(n - 1)
  print(s)
  return s
print(sumofn(5))
```

```python
def fib(n):
  if n<=1:
    return n
  return fib(n-1)+fib(n-2)
fib(4)
```

```python
# print all subsets
def subsets(i,nums,l):
  if i==len(nums):
    print(l)
    return
  l.append(nums[i])
  subsets(i+1,nums,l)
  l.pop()
  subsets(i+1,nums,l)

subsets(0,[1,2,3],[])
```
