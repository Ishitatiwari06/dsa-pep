```python
# linear search
def linsearch(arr,target):
  for i,j in enumerate(arr):
    if(j==target):
      return i
  return -1
print(linsearch([1,2,3,4,5],6))
```

```python
# binary search
def binsearch(l, r, arr, target):
  while l <= r:
    mid = (l + r) // 2
    if arr[mid] == target:
      return mid
    elif arr[mid] < target:
      l = mid + 1
    else:
      r = mid - 1
  return -1

arr = [1, 2, 3, 4, 5, 6]
print(binsearch(0, len(arr) - 1, arr, 4))


```
