```python
class Node:
  def __init__(self,data):
    self.data=data
    self.next=None
```

```python
def display(head):
  cur=head
  while(cur):
    print(cur.data)
    cur=cur.next
```

```python
def insert_beg(head,val):
  newNode=Node(val)
  newNode.next=head
  head=newNode
  return head
def insert_end(tail,val):
  newNode=Node(val)
  tail.next=newNode
  tail=tail.next
  return tail
```

```python
head=Node(1)
tail=Node(2)
head.next=tail
head=insert_beg(head,3)
tail=insert_end(tail,4)
display(head)
```

```python
def search(head,val):
  n=head
  while n:
    if n.data==val:
      return True
  return False
```

```python
def delete_last(head,tail):
  node=head
  if((not head) or head==tail):
    return None
  while(node.next!=tail):
    node=node.next
  node.next=None
  tail=node
  return head
print("Before Delete")
display(head)
delete_last(head,tail)
print("After Delete")
display(head)
```

```python
# length of linked list
def length(head):
  cnt=0
  cur=head
  while(cur):
    cur=cur.next
    cnt+=1
  return cnt

print(length(head))
```
